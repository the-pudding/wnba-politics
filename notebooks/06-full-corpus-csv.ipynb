# One CSV with everything

Concatenates every per-outlet CSV in `data/` into `data/all_outlets_full.csv` —
all ~59,000 articles, full body text where the outlet allows it. ~170MB: fine as a
Google Drive file, too big for Google Sheets (browse `all_outlets_index.csv` there instead).


```python
from pathlib import Path
import pandas as pd

DATA = Path("../data")
OUTLETS = ["fox_wnba", "fox_nba", "yahoo_sports", "espn",
           "just_womens_sports", "the_athletic", "cbs_sports", "usa_today"]

full = pd.concat((pd.read_csv(DATA / f"{o}.csv") for o in OUTLETS), ignore_index=True)
full.to_csv(DATA / "all_outlets_full.csv", index=False)
print(f"{len(full):,} rows, {(DATA / 'all_outlets_full.csv').stat().st_size / 1e6:.0f} MB")
full.outlet.value_counts()
```
