# The chyron ternary, in a table

`chyron_words.csv` holds every word from the three networks' WNBA
chyrons: how many chyrons it appears in, and each network's share of
those uses (shares are normalized so a bigger network doesn't win every
word by default). A word "belongs" to the network with the biggest share.

```python
import pandas as pd

words = pd.read_csv("../data/chyron_words.csv")
words.sample(5)
```

```python
# each network's most distinctive words (min 25 chyrons)
solid = words[words.total_chyrons >= 25]
for channel in [c for c in words.columns if c.startswith("share_")]:
    name = channel.replace("share_", "").replace("_", " ").title()
    top = solid.sort_values(channel, ascending=False).head(15)
    print(f"\n=== most {name} ===")
    print(top[["word", "total_chyrons", channel]].to_string(index=False))
```

```python
# the piece's example: the Griner swap, told two ways
swap = words[words.word.isin(["viktor", "bout", "merchant", "death", "trade"])]
print(swap.to_string(index=False))
```

MSNBC named the man (*viktor*, *bout*); Fox named the myth
(*merchant*, *death*). Same swap, two vocabularies.
