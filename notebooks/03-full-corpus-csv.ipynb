# One CSV with everything

Output files:
`data/all_outlets_full.csv` (gitignored — too big for GitHub).

Input files: every per-outlet CSV in `data/`.


## Setup

```python
from pathlib import Path
import pandas as pd

DATA = Path("../data")
import sys as _sys; _sys.path.insert(0, ".")
from outlets import OUTLETS as _OUTLETS
OUTLETS = [key for key, *_ in _OUTLETS]   # the registry's outlets (bundled files included)
# the Fox NBA control corpus rides along for the timeline's NBA comparison
OUTLETS += [k for k in ("fox_nba",) if (DATA / f"{k}.csv").exists()]

full = pd.concat((pd.read_csv(DATA / f"{o}.csv") for o in OUTLETS), ignore_index=True)
```

#### Overwrites `data/all_outlets_full.csv`

```python
full.to_csv(DATA / "all_outlets_full.csv", index=False)
print(f"{len(full):,} rows, {(DATA / 'all_outlets_full.csv').stat().st_size / 1e6:.0f} MB")
full.outlet.value_counts()
```

## Spot check

```python
for _, r in full.sample(5).iterrows():
    body = (r.body or "")[:70] if isinstance(r.body, str) else "(headline only)"
    print(f"   [{r.outlet}] {str(r.date)[:10]} {str(r.headline)[:64]}")
    print(f"      {body}")
```
